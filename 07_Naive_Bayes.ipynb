# Module 7 — Naive Bayes

Naive Bayes is a fast, simple classifier based on **probability**.

We'll use the Hotel Bookings dataset for the numerical example, and a small set of text messages for the text classification example.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score
```

### Load the Data

We'll use the Hotel Bookings dataset.

The target is `is_canceled`:

- `1` = Canceled
- `0` = Not Canceled

```python
data = pd.read_csv("data/hotel_bookings.csv")

data["children"] = data["children"].fillna(0)

data = data[
    [
        "lead_time",
        "adults",
        "children",
        "babies",
        "adr",
        "total_of_special_requests",
        "is_canceled"
    ]
].dropna()

data = data.sample(200, random_state=1).reset_index(drop=True)

data[
    [
        "lead_time",
        "adr",
        "is_canceled"
    ]
].head()
```

## 1. Bayes Theorem

**Bayes Theorem** is a formula for updating a probability once you see new evidence.

It combines what you already know with what the new data tells you.

## 2. Conditional Probability

**Conditional probability** is the chance of something happening, **given** that something else has already happened.

Example: the probability that a booking is canceled, **given** that its lead time is greater than 100 days.

```python
long_lead_bookings = data[data["lead_time"] > 100]

prob_canceled_given_long_lead = long_lead_bookings["is_canceled"].mean()

print(
    "P(Canceled | Lead Time > 100):",
    round(prob_canceled_given_long_lead, 2)
)
```

## 3. Naive Bayes

**Naive Bayes** is a classifier that uses Bayes Theorem to calculate the probability of each class using the available features.

It then predicts the class with the highest probability.

It is popular because it is fast, simple, and works well for many classification problems, especially text classification.

### Why is it called "Naive"?

It is called **"Naive"** because it makes a simple independence assumption.

The model assumes that the features are independent of each other when calculating the probability.

This assumption is often not completely true in real-world data, but it makes the probability calculation simple and fast.

Even with this simplified assumption, Naive Bayes can perform well on many problems.

## 4. Independence Assumption

Naive Bayes assumes that all features are **independent** of each other.

For example, it assumes that `lead_time` and `adr` do not affect each other when calculating the probability of cancellation.

In real-world data, features can be related, but Naive Bayes treats them independently to simplify the calculation.

## 5. Gaussian Naive Bayes

**Gaussian Naive Bayes** is used when the features are **continuous numerical values**.

Examples in our Hotel Bookings dataset include:

- `lead_time`
- `adr`
- `adults`
- `total_of_special_requests`

We'll use these features to predict whether a booking is canceled.

```python
X = data[
    [
        "lead_time",
        "adults",
        "children",
        "babies",
        "adr",
        "total_of_special_requests"
    ]
]

y = data["is_canceled"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=1,
    stratify=y
)

gnb = GaussianNB()

gnb.fit(X_train, y_train)

predictions = gnb.predict(X_test)

print(
    "Accuracy:",
    round(accuracy_score(y_test, predictions), 2)
)
```

## 6. Text Classification Use Case

Naive Bayes is one of the common choices for **text classification**, such as spam detection.

Text can be converted into numerical word features and then classified using Naive Bayes.

Let's build a small spam-detection example.

```python
messages = [
    "win a free prize now",
    "claim your free lottery winnings",
    "meeting scheduled for tomorrow",
    "please review the attached report",
    "you have won a free gift card",
    "let's catch up over lunch",
    "urgent: claim your reward now",
    "here are the notes from today's meeting",
]

labels = [1, 1, 0, 0, 1, 0, 1, 0]

# 1 = spam
# 0 = not spam
```

## 7. Multinomial Naive Bayes

**Multinomial Naive Bayes** works with **counts**, such as how many times each word appears in a message.

It is commonly used for text classification.

```python
vectorizer = CountVectorizer()

word_counts = vectorizer.fit_transform(messages)

mnb = MultinomialNB()

mnb.fit(word_counts, labels)

new_message = ["free prize waiting for you"]

new_counts = vectorizer.transform(new_message)

prediction = mnb.predict(new_counts)

print(
    "Prediction (1=spam, 0=not spam):",
    prediction[0]
)
```

## 8. Bernoulli Naive Bayes

**Bernoulli Naive Bayes** only cares whether a word appears or not.

It treats each word as:

- `1` = word is present
- `0` = word is absent

It can work well for short text where word presence is more important than the number of times a word appears.

```python
binary_vectorizer = CountVectorizer(binary=True)

binary_counts = binary_vectorizer.fit_transform(messages)

bnb = BernoulliNB()

bnb.fit(binary_counts, labels)

new_binary = binary_vectorizer.transform(new_message)

prediction = bnb.predict(new_binary)

print(
    "Prediction (1=spam, 0=not spam):",
    prediction[0]
)
```
