## 1. What is KNN?

**K-Nearest Neighbors (KNN)** predicts something about a new point by looking at the **K closest points** to it in the existing data.

For classification, the neighbors vote on the class.

For regression, their target values are averaged.

KNN does not learn a traditional mathematical equation during training. Instead, it stores the training data and performs the comparison when making predictions.

That's why KNN is called a **lazy learner**.

## 2. Distance

To find the "nearest" neighbors, KNN needs a way to measure how **close** two points are.

This measurement is called **distance**.

The smaller the distance, the more similar the two observations are considered.

## 3. Euclidean Distance

The most common distance measure is **Euclidean Distance**.

It is the straight-line distance between two points.

For two points:

`(x1, y1)` and `(x2, y2)`

the distance is:

`√((x2 - x1)² + (y2 - y1)²)`

```python
point_a = np.array([2, 3])
point_b = np.array([5, 7])

distance = np.sqrt(np.sum((point_a - point_b) ** 2))

print("Euclidean distance:", distance)
```

## 4. Neighbors

The **neighbors** of a point are the other data points that are closest to it based on the selected distance measurement.

KNN looks at the K nearest neighbors and uses them to make its prediction.

For example, when `K = 5`, the model looks at the 5 closest observations.

## 5. Feature Scaling

KNN compares distances between observations.

Features with larger numerical ranges can dominate the distance calculation.

For example:

- `lead_time` can have values in the hundreds
- `babies` usually has much smaller values

**Feature Scaling** puts the features on a comparable scale before calculating distances.

We'll use `StandardScaler`.

## 6. Classification with KNN

For **classification**, KNN looks at the K nearest neighbors and predicts whichever class is the **majority** among them.

We'll predict `is_canceled` using the selected Hotel Booking features.

- `0` = Not Canceled
- `1` = Canceled

## 7. Choosing K

**K** is the number of neighbors the model looks at.

Choosing the right K is important.

- **Small K** such as 1 → very sensitive to noise and can overfit
- **Large K** → smoother predictions, but may miss important local patterns

We can try different K values and compare their performance.

## 8. Regression with KNN

KNN is not only used for classification. It can also predict **continuous numbers**.

For **KNN Regression**, instead of voting on a class, the model averages the target values of the K nearest neighbors.

## 9. Advantages

- Simple to understand and easy to implement
- No complex training process
- Works for both classification and regression
- Can naturally handle multiple classes
- Can model nonlinear relationships

## 10. Limitations

- Slow on large datasets because it compares new observations with training observations
- Needs feature scaling to work well
- Sensitive to irrelevant features
- Sensitive to noisy data
- Choosing the wrong K can affect performance
- Requires storing the training data
