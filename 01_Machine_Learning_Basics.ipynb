# Module 1 - Introduction to Machine Learning

## 1. What is Machine Learning?

Machine Learning (ML) is a branch of Artificial Intelligence in which a system learns patterns from data and uses those patterns to make predictions or decisions without being explicitly programmed with fixed rules.

Example:

In a retail dataset, instead of manually writing rules to predict customer behaviour, an ML model can learn patterns from previous transactions.

---

## 2. Why Machine Learning?

Machine Learning is useful because:

- Some problems are too complex to solve using fixed rules.
- ML can identify patterns in large datasets.
- Models can improve when more relevant data becomes available.
- ML can automate prediction and decision-making tasks.

Examples:

- Fraud detection
- Customer churn prediction
- Product recommendations
- Sales forecasting
- Spam detection

---

## 3. AI vs ML vs Deep Learning

| Concept | Description |
|---|---|
| Artificial Intelligence (AI) | Broad field of creating systems that perform tasks requiring human-like intelligence. |
| Machine Learning (ML) | A subset of AI where systems learn patterns from data. |
| Deep Learning (DL) | A subset of ML that uses multi-layer neural networks. |

Relationship:

`AI → Machine Learning → Deep Learning`

---

## 4. Traditional Programming vs Machine Learning

| Traditional Programming | Machine Learning |
|---|---|
| Rules + Data → Output | Data + Output → Model |
| Human writes the rules | Model learns patterns from data |
| Suitable for well-defined rules | Suitable for complex pattern-based problems |

Example:

For spam detection, traditional programming requires manually created rules. In Machine Learning, the model learns from examples of spam and non-spam emails.

## 6. Supervised Learning

In Supervised Learning, the model learns from labelled data.

The model learns a relationship between:

`Features → Target`

Example:

Using hotel booking features to predict whether a booking will be cancelled.

```text
Features:
lead_time
adr
adults
children
total_of_special_requests

Target:
is_canceled
```

Common supervised learning tasks:

- Classification
- Regression

---

## 7. Unsupervised Learning

In Unsupervised Learning, the data does not contain a predefined target.

The model tries to find hidden patterns or structures in the data.

Example:

Grouping hotel bookings or customers based on similar characteristics.

Common task:

- Clustering

---

## 8. Semi-Supervised Learning

Semi-Supervised Learning uses both labelled and unlabelled data.

It is useful when obtaining labelled data is expensive or time-consuming.

Example:

A small portion of booking records has labels, while a large portion does not.

---

## 9. Reinforcement Learning

Reinforcement Learning is based on interaction between an agent and an environment.

The agent:

1. Takes an action.
2. Receives a reward or penalty.
3. Learns from the result.
4. Improves its future actions.

Example:

An agent learning how to choose actions to maximize a reward.

---

## 10. Classification

Classification is a supervised learning task where the target is a category or class.

Example:

Predict whether a hotel booking will be cancelled.

```text
0 → Not Cancelled
1 → Cancelled
```

```python
df["is_canceled"].value_counts()
```

A classification model can learn from the booking features and predict the class for new bookings.

---

## 11. Regression

Regression is a supervised learning task where the target is a continuous numerical value.

Example:

Predicting the Average Daily Rate (`adr`) using other booking features.

```python
df["adr"].describe()
```

A regression model produces a numerical prediction.

---

## 12. Clustering

Clustering is an unsupervised learning technique that groups similar observations together without predefined labels.

Example:

Hotel bookings can potentially be grouped based on characteristics such as:

- Lead time
- Number of guests
- Stay duration
- ADR

Later, K-Means Clustering will be used to explore this concept in detail.

# 13. Core Machine Learning Terminology

## 13.1 Features

Features are the input variables used by the model to make predictions.

Example:

```text
lead_time
adults
children
adr
total_of_special_requests
```

---

## 13.2 Target

The target is the output variable that the model tries to predict.

Example:

```text
is_canceled
```

---

## 13.3 Training

Training is the process of providing data to a model so that it can learn patterns.

Example:

```python
model.fit(X_train, y_train)
```

---

## 13.4 Prediction

Prediction means using a trained model to generate an output for new data.

```python
model.predict(X_test)
```

---

## 13.5 Model

A model is the mathematical representation learned from the training data.

Examples:

```python
LinearRegression()
LogisticRegression()
KMeans()
```

---

## 13.6 Parameters

Parameters are values that the model learns automatically during training.

Example:

In Linear Regression, coefficients and intercept are learned from the training data.

---

## 13.7 Hyperparameters

Hyperparameters are settings specified before or during model training. They are not directly learned from the training data.

Examples:

```text
K in KNN
C in SVM
n_clusters in K-Means
```
